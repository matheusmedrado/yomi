# Yomi — experimento de PDI para OCR de mangá

Este notebook é a trilha acadêmica do projeto. A pergunta é: **aplicar PDI clássica ao mesmo recorte melhora o OCR japonês?**

O detector e o OCR são ferramentas externas. O que avaliamos é o trecho entre eles: tons de cinza, CLAHE, filtragem, Otsu, morfologia, componentes conectados e recorte por projeção. Frontend, tradução, dicionário, furigana e Anki pertencem à aplicação, mas estão fora da contribuição de PDI.

## 1. Preparação

Execute o notebook a partir da raiz do repositório ou da pasta `notebooks`. Os modelos do detector e do `manga-ocr` precisam estar disponíveis localmente para a avaliação completa.

```python
from pathlib import Path
import json
import sys

import cv2
import numpy as np
from PIL import Image

ROOT = Path.cwd().resolve()
if not (ROOT / 'backend').is_dir():
    ROOT = ROOT.parent
if not (ROOT / 'backend').is_dir():
    raise RuntimeError('Abra o notebook a partir da raiz do projeto ou da pasta notebooks.')
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from backend.ocr import MangaOcrService
from backend.pipeline.conditioning import (
    _clean_mask, _component_geometry, _ink_mask, condition_crop, normalize_crop,
)
from backend.pipeline.detection import detect_blocks, get_detector
from backend.pipeline.metrics import evaluate_regions
from backend.pipeline.preprocess import clahe_equalize, denoise, to_grayscale

PAGE_PATH = ROOT / 'sample' / '08.png'
TRUTH_PATH = ROOT / 'backend' / 'data' / 'ground_truth' / 'sample_08_11_transcripts.json'
print('Raiz:', ROOT)
print('Página de exemplo:', PAGE_PATH)
```

## 2. Página real e regiões candidatas

A localização abaixo vem do `comic-text-detector`. Isso não é apresentado como implementação do grupo. A comparação Baseline × PDI reutilizará exatamente essas mesmas caixas e linhas.

```python
page = cv2.imread(str(PAGE_PATH), cv2.IMREAD_COLOR)
if page is None:
    raise FileNotFoundError(PAGE_PATH)

detector = get_detector(device='cpu')
if not detector.available:
    raise RuntimeError('Modelo comictextdetector.pt não encontrado.')

# Baseline devolve os recortes do detector sem aplicar condition_crop.
blocks = detect_blocks(page, detector=detector, mode='baseline')
overlay = page.copy()
for block in blocks:
    cv2.rectangle(overlay, (block.x, block.y),
                  (block.x + block.w, block.y + block.h), (0, 0, 220), 3)
    cv2.putText(overlay, str(block.id), (block.x, max(20, block.y - 5)),
                cv2.FONT_HERSHEY_SIMPLEX, 0.7, (0, 0, 220), 2, cv2.LINE_AA)

print(f'{len(blocks)} regiões; {sum(len(b.crops) for b in blocks)} linhas candidatas')
Image.fromarray(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
```

## 3. Núcleo PDI inspecionável

O recorte de uma linha passa pelas seguintes operações:

1. BGR → tons de cinza;
2. CLAHE para contraste local;
3. filtro bilateral para reduzir ruído preservando transições;
4. Otsu normal e invertido, escolhendo uma polaridade plausível para a tinta;
5. abertura e fechamento morfológicos;
6. componentes conectados para validar e limitar a região útil;
7. perfil de projeção para cortar linhas excessivamente longas em vales de pouca tinta.

CLAHE e o filtro bilateral são extensões, respectivamente, da equalização do Lab 03 e da filtragem passa-baixa do Lab 04. Otsu, morfologia e componentes conectados correspondem diretamente aos Labs 06, 07 e 02.

```python
def _as_bgr(image):
    if image.ndim == 2:
        return cv2.cvtColor(image, cv2.COLOR_GRAY2BGR)
    return image.copy()

def _tile(label, image, width=320, height=220):
    image = _as_bgr(image)
    canvas = np.full((height, width, 3), 255, dtype=np.uint8)
    usable_w, usable_h = width - 20, height - 45
    scale = min(usable_w / image.shape[1], usable_h / image.shape[0])
    new_w = max(1, int(round(image.shape[1] * scale)))
    new_h = max(1, int(round(image.shape[0] * scale)))
    resized = cv2.resize(image, (new_w, new_h), interpolation=cv2.INTER_AREA)
    x, y = (width - new_w) // 2, 35 + (usable_h - new_h) // 2
    canvas[y:y + new_h, x:x + new_w] = resized
    cv2.putText(canvas, label, (10, 24), cv2.FONT_HERSHEY_SIMPLEX,
                0.58, (30, 30, 30), 1, cv2.LINE_AA)
    return canvas

raw_crop = blocks[0].crops[0]
result = condition_crop(raw_crop)
final_crop = np.concatenate(result.crops, axis=1)
tiles = [
    _tile('1. Recorte original', result.raw),
    _tile('2. Cinza + CLAHE + bilateral', result.enhanced),
    _tile('3. Otsu', result.mask),
    _tile('4. Componentes e limites', result.components_overlay),
    _tile('5. Perfil de projeção', result.projection),
    _tile('6. Imagem entregue ao OCR', final_crop),
]
panel = np.vstack((np.hstack(tiles[:3]), np.hstack(tiles[3:])))
print('Fallback:', result.fallback or 'nenhum', '| cortes:', result.cut_points)
Image.fromarray(cv2.cvtColor(panel, cv2.COLOR_BGR2RGB))
```

## 4. Avaliação justa: a mesma detecção nos dois caminhos

Para não confundir localização com condicionamento, o detector é executado uma vez. Cada linha original alimenta o Baseline; a mesma linha passa por `condition_crop` e alimenta o caminho PDI.

Relatamos separadamente:

- **cobertura da localização**: proporção das anotações associadas a alguma caixa;
- **CER ponta a ponta**: inclui anotações não localizadas como erro total;
- **CER nas associadas**: compara o texto somente onde houve associação geométrica.

A terceira medida é a mais direta para responder se o condicionamento alterou o OCR; a primeira mede o detector externo.

```python
truth_data = json.loads(TRUTH_PATH.read_text(encoding='utf-8'))
truth_pages = truth_data.get('pages', truth_data)
ocr = MangaOcrService.instance()

def scaled_truth(page_data, image):
    sx = image.shape[1] / float(page_data.get('annotation_width', image.shape[1]))
    sy = image.shape[0] / float(page_data.get('annotation_height', image.shape[0]))
    boxes = page_data.get('rough_boxes', {})
    output = []
    for region in page_data.get('regions', []):
        box = boxes.get(region.get('bubble_id'))
        if box is None:
            continue
        x, y, w, h = box
        output.append({**region, 'x': round(x * sx), 'y': round(y * sy),
                       'w': round(w * sx), 'h': round(h * sy)})
    return output

def metric_summary(mode, metrics, detected, fallback_count=0):
    associated = [metric for metric in metrics if metric.prediction_id is not None]
    total = len(metrics)
    return {
        'modo': mode,
        'caixas_detectadas': detected,
        'anotacoes_associadas': f'{len(associated)}/{total}',
        'cobertura_localizacao': round(len(associated) / max(1, total), 3),
        'CER_ponta_a_ponta': round(sum(m.cer for m in metrics) / max(1, total), 3),
        'CER_apenas_associadas': (
            round(sum(m.cer for m in associated) / len(associated), 3)
            if associated else None
        ),
        'fallbacks_normalizados': fallback_count,
    }

def evaluate_same_blocks(image, page_truth, detected_blocks=None):
    if detected_blocks is None:
        detected_blocks = detect_blocks(image, detector=detector, mode='baseline')
    baseline_predictions = []
    pdi_predictions = []
    fallback_count = 0

    for block in detected_blocks:
        baseline_text = ''.join(ocr.recognize(crop) for crop in block.crops)
        conditioned_crops = []
        for crop in block.crops:
            conditioned = condition_crop(crop)
            fallback_count += conditioned.fallback == 'normalized'
            conditioned_crops.extend(conditioned.crops)
        pdi_text = ''.join(ocr.recognize(crop) for crop in conditioned_crops)
        geometry = block.to_dict()
        baseline_predictions.append({**geometry, 'text': baseline_text})
        pdi_predictions.append({**geometry, 'text': pdi_text})

    expected = scaled_truth(page_truth, image)
    baseline_metrics = evaluate_regions(baseline_predictions, expected)
    pdi_metrics = evaluate_regions(pdi_predictions, expected)
    return [
        metric_summary('Baseline', baseline_metrics, len(detected_blocks)),
        metric_summary('PDI', pdi_metrics, len(detected_blocks), fallback_count),
    ]

truth_08 = next(item for item in truth_pages if item.get('page') == PAGE_PATH.name)
page_08_results = evaluate_same_blocks(page, truth_08, blocks)
page_08_results
```

## 5. Repetição nas quatro páginas anotadas

Uma página isolada é apenas um exemplo. A célula seguinte repete exatamente o mesmo protocolo nas páginas 08–11. Ela pode levar alguns minutos em CPU porque executa o OCR nos dois caminhos.

```python
all_results = []
page_blocks_cache = {}
for page_truth in truth_pages:
    page_name = page_truth.get('page')
    image = cv2.imread(str(ROOT / 'sample' / page_name), cv2.IMREAD_COLOR)
    if image is None:
        print('Ignorando página ausente:', page_name)
        continue
    print('Avaliando', page_name, '...')
    detected_for_page = detect_blocks(image, detector=detector, mode='baseline')
    page_blocks_cache[page_name] = (image, detected_for_page)
    rows = evaluate_same_blocks(image, page_truth, detected_for_page)
    for row in rows:
        all_results.append({'pagina': page_name, **row})

all_results
```

## 6. Resultado agregado

A média agregada é ponderada pela quantidade de anotações de cada página. Isso evita escolher apenas uma página favorável ao método.

```python
def aggregate_rows(rows):
    output = []
    for mode in sorted({row['modo'] for row in rows}):
        selected = [row for row in rows if row['modo'] == mode]
        pairs = [tuple(map(int, row['anotacoes_associadas'].split('/'))) for row in selected]
        associated = sum(pair[0] for pair in pairs)
        total = sum(pair[1] for pair in pairs)
        end_to_end = sum(row['CER_ponta_a_ponta'] * pair[1]
                            for row, pair in zip(selected, pairs)) / max(1, total)
        matched_values = [
            (row['CER_apenas_associadas'], pair[0])
            for row, pair in zip(selected, pairs)
            if row['CER_apenas_associadas'] is not None
        ]
        matched_cer = sum(value * count for value, count in matched_values) / max(1, associated)
        output.append({
            'modo': mode,
            'anotacoes_associadas': f'{associated}/{total}',
            'cobertura_localizacao': round(associated / max(1, total), 3),
            'CER_ponta_a_ponta': round(end_to_end, 3),
            'CER_apenas_associadas': round(matched_cer, 3),
        })
    return output

aggregate_rows(all_results)
```

## 7. Ablação: qual etapa realmente ajuda?

Aplicar todas as técnicas disponíveis não torna um pipeline automaticamente melhor. A ablação abaixo mantém as mesmas caixas e compara entradas progressivamente transformadas. Ela também inclui equalização global, Gaussiano e mediana, que correspondem mais diretamente aos laboratórios.

Esta célula é opcional e mais demorada: cada variante executa novamente o OCR nas quatro páginas.

```python
def _gray_bgr(gray):
    return cv2.cvtColor(gray, cv2.COLOR_GRAY2BGR)

def pdi_crop_with_original_pixels(raw, grayscale=False):
    enhanced = normalize_crop(raw)
    mask = _clean_mask(_ink_mask(enhanced))
    valid, bounds, _overlay = _component_geometry(mask)
    if not valid or bounds is None:
        return [_gray_bgr(to_grayscale(raw))] if grayscale else [raw]
    x0, y0, x1, y1 = bounds
    source = _gray_bgr(to_grayscale(raw)) if grayscale else raw
    return [source[y0:y1, x0:x1]]

def variant_crops(raw, variant):
    gray = to_grayscale(raw)
    if variant == 'Baseline':
        return [raw]
    if variant == 'Somente cinza':
        return [_gray_bgr(gray)]
    if variant == 'Equalizacao global':
        return [_gray_bgr(cv2.equalizeHist(gray))]
    enhanced = clahe_equalize(gray)
    if variant == 'CLAHE':
        return [_gray_bgr(enhanced)]
    if variant == 'CLAHE + Gaussiano':
        return [_gray_bgr(denoise(enhanced, method='gaussian'))]
    if variant == 'CLAHE + mediana':
        return [_gray_bgr(denoise(enhanced, method='median'))]
    if variant == 'CLAHE + bilateral':
        return [_gray_bgr(denoise(enhanced, method='bilateral'))]
    if variant == 'Recorte PDI + pixels originais':
        return pdi_crop_with_original_pixels(raw)
    if variant == 'Recorte PDI + cinza':
        return pdi_crop_with_original_pixels(raw, grayscale=True)
    if variant == 'Pipeline completo':
        return condition_crop(raw).crops
    raise ValueError(variant)

ABLATION_VARIANTS = [
    'Baseline', 'Somente cinza', 'Equalizacao global', 'CLAHE',
    'CLAHE + Gaussiano', 'CLAHE + mediana', 'CLAHE + bilateral',
    'Recorte PDI + pixels originais', 'Recorte PDI + cinza',
    'Pipeline completo',
]

ablation_rows = []
for page_truth in truth_pages:
    page_name = page_truth.get('page')
    if page_name not in page_blocks_cache:
        continue
    image, detected_blocks = page_blocks_cache[page_name]
    expected = scaled_truth(page_truth, image)
    print('Ablação', page_name, '...')
    predictions = {variant: [] for variant in ABLATION_VARIANTS}
    for block in detected_blocks:
        for variant in ABLATION_VARIANTS:
            text = ''
            for raw in block.crops:
                text += ''.join(ocr.recognize(crop)
                                for crop in variant_crops(raw, variant))
            predictions[variant].append({**block.to_dict(), 'text': text})
    for variant in ABLATION_VARIANTS:
        metrics = evaluate_regions(predictions[variant], expected)
        row = metric_summary(variant, metrics, len(detected_blocks))
        ablation_rows.append({'pagina': page_name, **row})

sorted(aggregate_rows(ablation_rows), key=lambda row: row['CER_apenas_associadas'])
```

## 8. Como interpretar sem exagerar

- Baseline e PDI têm necessariamente a mesma cobertura, pois usam a mesma detecção.
- CER menor é melhor. A contribuição do condicionamento deve ser discutida principalmente pelo `CER_apenas_associadas`.
- Cobertura baixa indica limitação das caixas do detector ou das anotações aproximadas; não deve ser atribuída ao condicionamento.
- Se a melhora variar entre páginas, mostre essa variação. Não transforme um ganho pequeno em uma alegação de precisão geral.
- Os estágios visuais explicam **como** o PDI modifica a entrada; a tabela avalia **se** a modificação ajudou o OCR.

A aplicação completa pode ser demonstrada depois deste experimento, deixando claro que tradução, dicionário, furigana, Anki e frontend são extensões autorizadas fora do escopo acadêmico.
